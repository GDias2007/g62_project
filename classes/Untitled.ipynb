```python
import pandas as pd
import sqlite3


df = pd.read_csv('g62_Banks_Cards.csv', sep=';')


conn = sqlite3.connect('data/g62_database.db')


df_banks = df[['bank_id', 'designation', 'founding_date']].drop_duplicates()
df_banks.to_sql('Banks', conn, if_exists='append', index=False)


df_cards = df[['card_id', 'name', 'card_type']].drop_duplicates()
df_cards.to_sql('Cards', conn, if_exists='append', index=False)


df_branches = df[['branch_id', 'branch_location', 'bank_id']].drop_duplicates().dropna(subset=['branch_id'])
df_branches.to_sql('Branches', conn, if_exists='append', index=False)


df_trans = df[['transaction_date', 'amount', 'card_id']]
df_trans.to_sql('Transactions', conn, if_exists='append', index=False)

df_bankcard = df[['bank_id', 'card_id']].drop_duplicates()
df_bankcard.to_sql('BankCards', conn, if_exists='append', index=False)


conn.close()
print("Sucesso! Os dados foram convertidos para a Base de Dados.")
```

```python

banks = df[['bank_id', 'designation', 'founding_date']].drop_duplicates()
banks.to_sql('Banks', conn, if_exists='replace', index=False)
print("Tabela Banks criada!")
```

```python

branches = df[['branch_id', 'branch_location', 'bank_id']].drop_duplicates()
branches = branches.dropna(subset=['branch_id'])
branches.to_sql('Branches', conn, if_exists='replace', index=False)
print("Tabela Branches criada!")
```

```python
cards = df[['card_id', 'name', 'card_type']].drop_duplicates()
cards.to_sql('Cards', conn, if_exists='replace', index=False)
print("Tabela Cards criada!")
```

```python
transactions = df[['transaction_date', 'amount', 'card_id']]
transactions.to_sql('Transactions', conn, if_exists='replace', index=False)
print("Tabela Transactions criada!")
```

```python
conn.close()
print("Base de dados guardada e fechada!")
```
